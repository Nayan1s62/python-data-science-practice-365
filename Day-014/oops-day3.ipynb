```python
from abc import ABC, abstractmethod


# Abstract Parent Class
class Payment(ABC):

    # Constructor
    def __init__(self, owner, amount):
        self.__Owner_name = owner
        self.__Amount = amount

    # Getter for Owner Name
    def get_owner(self):
        return f"Owner Name: {self.__Owner_name}"

    # Getter for Amount
    def get_amount(self):
        return f"Amount: {self.__Amount}"

    # Setter for Amount
    def set_amount(self, amt):
        if amt > 0:
            self.__Amount += amt
            return f"Amount Updated: {self.__Amount}"

        else:
            return "Please Enter Amount More Than Zero"

    # Abstract Method
    @abstractmethod
    def pay(self):
        pass

    # Instance Method
    def receipt(self):
        print(f"Owner Name: {self.__Owner_name}")
        print(f"Amount: {self.__Amount}")


# Child Class 1
class CreditCard(Payment):

    def __init__(self, owner, amount, credit_card_no):
        super().__init__(owner, amount)
        self.Credit_card = credit_card_no

    def pay(self):
        print("Credit Card Payment Details:")
        print(f"Credit Card Number: {self.Credit_card}")
        self.receipt()


# Child Class 2
class DebitCard(Payment):

    def __init__(self, owner, amount, debit_card_no):
        super().__init__(owner, amount)
        self.Debit_card = debit_card_no

    def pay(self):
        print("Debit Card Payment Details:")
        print(f"Debit Card Number: {self.Debit_card}")
        self.receipt()


# Child Class 3
class UPI(Payment):

    def __init__(self, owner, amount, upi_id):
        super().__init__(owner, amount)
        self.UPI_ID = upi_id

    def pay(self):
        print("UPI Payment Details:")
        print(f"UPI ID: {self.UPI_ID}")
        self.receipt()


# Creating Objects

credit = CreditCard("Rohit", 10000, 9420930001)

debit = DebitCard("Yash", 20000, 899991000)

upi = UPI("Gokul", 5000, "gokul@upi")


# Calling Methods

credit.pay()

print("----------------")

debit.pay()

print("----------------")

upi.pay()

print("----------------")

print(credit.set_amount(5000))
```

```python
from abc import ABC, abstractmethod

# Abstract Class
class Agent(ABC):

    # Constructor
    def __init__(self, agent_name, agent_apikey):
        self.agent_name = agent_name
        self.__agent_apikey = agent_apikey

    # Getter
    def get_apikey(self):
        return self.__agent_apikey

    # Setter
    def set_apikey(self, new_key):
        self.__agent_apikey = new_key

    # Abstract Method
    @abstractmethod
    def llm_name(self):
        pass

    # Instance Method
    def display_info(self):
        print("Agent Name :", self.agent_name)
        print("Agent API Key :", self.__agent_apikey)


# Child Class 1
class FlightAgent(Agent):

    def __init__(self, agent_name, agent_apikey):
        super().__init__(agent_name, agent_apikey)

    def llm_name(self):
        print("LLM Name : GPT-4o (Flight Assistant)")

    def display_info(self):
        print("\n----- Flight Agent -----")
        super().display_info()
        self.llm_name()


# Child Class 2
class HotelBookingAgent(Agent):

    def __init__(self, agent_name, agent_apikey):
        super().__init__(agent_name, agent_apikey)

    def llm_name(self):
        print("LLM Name : Gemini Pro (Hotel Booking Assistant)")

    def display_info(self):
        print("\n----- Hotel Booking Agent -----")
        super().display_info()
        self.llm_name()


# Create Objects
flight = FlightAgent("Flight AI", "FL123API")
hotel = HotelBookingAgent("Hotel AI", "HT456API")

# Display Information
flight.display_info()
hotel.display_info()
```
