#####  Multi level inheritance  problem :



```python
# Grand Parent Class
class Person:

    # Constructor
    def __init__(self, name, age):
        self.__Name = name
        self.__Age = age

    # Getter for Name
    def get_name(self):
        return f"Name of Person: {self.__Name}"

    # Getter for Age
    def get_age(self):
        return f"Age of Person: {self.__Age}"

    # Setter for Age
    def set_age(self, age):
        if 20 < age < 60:
            self.__Age = age
            print(f"Age is updated: {self.__Age}")
        else:
            print("Please enter age between 20 and 60")

    # Person Information
    def person_info(self):
        return f"Name: {self.__Name}\nAge: {self.__Age}"


# Parent Class
class Employee(Person):

    # Constructor
    def __init__(self, name, age, salary, company):
        super().__init__(name, age)
        self.__Employee_Salary = salary
        self.__Company_Name = company

    # Getter for Salary
    def get_salary(self):
        return f"Salary of Employee: {self.__Employee_Salary}"

    # Getter for Company
    def get_company(self):
        return f"Company Name: {self.__Company_Name}"

    # Setter for Salary
    def set_salary(self, increment):

        if increment < (20 / 100) * self.__Employee_Salary:
            self.__Employee_Salary += increment
            print(f"Salary Updated: {self.__Employee_Salary}")

        else:
            print("Salary increment should be less than 20% of current salary")

    # Employee Information
    def employee_info(self):
        print("----- Employee Details -----")
        print(self.person_info())
        print(f"Salary: {self.__Employee_Salary}")
        print(f"Company: {self.__Company_Name}")


# Child Class
class Team(Employee):

    # Constructor
    def __init__(self, name, age, salary, company, team_size, bonus):
        super().__init__(name, age, salary, company)
        self.__Team_Size = team_size
        self.__Bonus = bonus

    # Getter for Team Size
    def get_team_size(self):
        return f"Team Size: {self.__Team_Size}"

    # Getter for Bonus
    def get_bonus(self):
        return f"Bonus: {self.__Bonus}"

    # Setter for Team Size
    def set_team_size(self, size):

        if size > 0:
            self.__Team_Size = size
            print(f"Team Size Updated: {self.__Team_Size}")

        else:
            print("Team size should be greater than zero")

    # Setter for Bonus
    def set_bonus(self, bonus):

        if bonus < (10 / 100) * self.__Bonus:
            self.__Bonus += bonus
            print(f"Bonus Updated: {self.__Bonus}")

        else:
            print("Bonus should be less than 10% of current bonus")

    # Team Information
    def team_info(self):
        print("----- Team Details -----")
        self.employee_info()
        print(f"Team Size: {self.__Team_Size}")
        print(f"Bonus: {self.__Bonus}")


# Main Program

team = Team(
    "Rahul",
    30,
    50000,
    "Google",
    10,
    5000
)

team.team_info()

print()

team.set_age(35)
team.set_salary(5000)
team.set_team_size(15)
team.set_bonus(200)
```
