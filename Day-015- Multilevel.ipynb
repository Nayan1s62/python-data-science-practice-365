```python
"""
Multilevel Inheritance Practice
"""

# Q1
class Animal:
    def species(self):
        print("Species: Animal")

class Dog(Animal):
    def breed(self):
        print("Breed: Labrador")

class Puppy(Dog):
    def puppy_name(self):
        print("Puppy Name: Bruno")

obj = Puppy()
obj.species()
obj.breed()
obj.puppy_name()


# Q2
class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def __init__(self, name, salary):
        super().__init__(name)
        self.salary = salary

class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)
        self.department = department

    def display(self):
        print(self.name, self.salary, self.department)

m = Manager("Nayan",70000,"IT")
m.display()


# Q3
class Vehicle:
    def __init__(self, company):
        self.company = company

class Car(Vehicle):
    def __init__(self, company, model):
        super().__init__(company)
        self.model = model

class ElectricCar(Car):
    def __init__(self, company, model, battery):
        super().__init__(company, model)
        self.battery = battery

    def display(self):
        print(self.company, self.model, self.battery)

ElectricCar("Tesla","Model 3","75 kWh").display()


# Constructor chaining
class A:
    def __init__(self):
        print("A Constructor")

class B(A):
    def __init__(self):
        super().__init__()
        print("B Constructor")

class C(B):
    def __init__(self):
        super().__init__()
        print("C Constructor")

C()
```
