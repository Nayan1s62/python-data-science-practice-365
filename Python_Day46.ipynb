Write a Python program to find the roots of a quadratic equation of the form:ax2+bx+c=0
The program should:
Accept values of a, b, and c as input from the user.
Calculate the discriminant using the formula:D = b² − 4ac
Use the quadratic formula to compute the roots.
Display both roots of the equation.


```python

import math

a = float(input())
b = float(input())
c = float(input())

D = b**2 - 4*a*c

root1 = (-b + math.sqrt(D)) / (2*a)
root2 = (-b - math.sqrt(D)) / (2*a)

print("Root 1 =", root1)
print("Root 2 =", root2)
```

Write a Python program to accept two numbers from the user.
Accept an arithmetic operator such as +, -, *, /, %, **, or //.
Perform the operation based on the given operator.
Calculate the result using arithmetic operators.
Print the final calculated output.

```python

a, op, b = input().split()

a = int(a)
b = int(b)

if op == "+":
    print(a + b)
elif op == "-":
    print(a - b)
elif op == "*":
    print(a * b)
elif op == "/":
    print(a / b)
elif op == "%":
    print(a % b)
elif op == "**":
    print(a ** b)
elif op == "//":
    print(a // b)
else:
    print("Invalid operator")

```

Write a Python program to read a string from the user.
Check each character to see if it is a vowel (a, e, i, o, u).
Use conditions to count only vowel characters.
Keep track of the total number of vowels.
Print the vowel count.


```python
text = input()
count = 0

for ch in text:
    if ch.lower() in "aeiou":
        count += 1

print(count)
```

```python
Write a Python program to read a string as input.
Remove duplicate characters while keeping the original order.
Process each character only once.
Create a new string without repeated characters.
Print the final modified string.
```

```python
text = input()
result = ""

for ch in text:
    if ch not in result:
        result += ch

print(result)
```

Write a Python program to perform basic list operations.
The program should:
Read a list of integers from the user.
Read a list operation command.
Perform one of the following operations based on the command:
   append x → add element x at the end of the list.
   insert i x → insert element x at index i.
   remove x → remove element x from the list.
Print the updated list after performing the operation.

```python
numbers = list(map(int, input().split()))
command = input().split()

if command[0] == "append":
    numbers.append(int(command[1]))

elif command[0] == "insert":
    numbers.insert(int(command[1]), int(command[2]))

elif command[0] == "remove":
    numbers.remove(int(command[1]))

print(numbers)
```

```python
Write a Python program with a function that accepts a list of numbers.
Calculate the maximum value using max().
Calculate the minimum value using min().
Compute the average of the numbers.
Return all three results as a tuple and print them.
```

```python
def calculate(numbers):
    maximum = max(numbers)
    minimum = min(numbers)
    average = sum(numbers) / len(numbers)

    return (maximum, minimum, average)

numbers = list(map(int, input().split()))

result = calculate(numbers)
print(result)
```
