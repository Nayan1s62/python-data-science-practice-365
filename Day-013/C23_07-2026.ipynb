```python
# payments 

from abc import ABC, abstractmethod

class Payement(ABC):

    # constructor :
    def __init__(self,owner,amount):
        self.__Owner_name = owner # private
        self.__Amount = int(amount) # private

    # getter : get the owner name
    def get_owner(self):
        return f"Owener Name : {self.__Owner_name}"

    # getter : get the amount
    def get_owner(self):
        return f"Amount  : {self.__Amount}"
    # setter
    def ser_amount(self,amt):
        
        if amt > 0:
            self.__Amount += amt
            return f"Amount updated : {self.__Amount}"

        else :
            print("Please enter the amount more than zero")

    # Abstarctmetod
    @abstractmethod
    def pay(self):
        pass

    #instance method
    def recipt(self):
        print(f"Owener Name : {self.__Owner_name}")
        print(f"Amount  : {self.__Amount}")


# child class
class Credit_card(Payement):

    def __init__(self,owner,amount,credit_card_no):
        super().__init__(owner,amount)
        self.__Credit_card = credit_card_no

    def pay(self):

        print("Credit card details :")
        print(f"Credit Card Number : {self.__Credit_card}")
        self.recipt()

class UPI_card(Payement):

    def __init__(self,owner,amount,upi_no):
        super().__init__(owner,amount)
        self.__Upi_card= upi_no

    def pay(self):

        print("UPI details :")
        print(f"UPI  Number : {self.__Upi_card}")
        self.recipt()
        
        

class Debit_card(Payement):

    def __init__(self,owner,amount,Debit_no):
        super().__init__(owner,amount)
        self.__Debit_card= Debit_no

    def pay(self):

        print("Debit card details :")
        print(f"UPI  Number : {self.__Debit_card}")
        self.recipt()
    
    
            
# creadit the calls object 

credit = Credit_card("Sumit",1000,12345)

debit = Debit_card("Yash",2000,234521)

Upi = UPI_card("Prashant",5000,3423433)

credit.pay()
print("#"*50)
debit.pay()
print("#"*50)
Upi.pay()


        
    

    
```

```python
# Agent  abstract class()
# three args : agent name, __agent_apikey

# abstract method : LLM_name 

# display all infor

# child 1: flight_book --> agent name, __agent_apikey, agent_role

# create LLM name abstract method logic

# display all info 

# child 2 : Hotel_booking --- Agent_nam , __Agent_apikey, agent_role

# create LLM name abstract methoop logic 

# display all info

from abc import ABC, abstractmethod

class Agent(ABC):

    # constructor :
    def __init__(self,agent_name,agent_apikey):
        self.Agent_name = agent_name # private
        self.__Agent_apikey = agent_apikey # private


    # getter : get the amount
    def get_agent_apikey(self):
        return f"Api key  : {self.__Agent_apikey}"
    # setter
    def ser_agent_apikey(self,key):
        
        if key > 0:
            self.__Agent_apikey= key
            return f"Agent name updated : {self.__Agent_apikey}"

        else :
            print("Please enter the valid Apikey ")

    # Abstarctmetod
    @abstractmethod
    def LLM_model(self):
        pass

    #instance method
    def Agent_info(self):
        print(f"Agen Name : {self.Agent_name}")
        print(f"APi key : {self.__Agent_apikey}")


# child class
class flight_book(Agent):

    def __init__(self,agent_name,agent_apikey,agent_role):
        super().__init__(agent_name,agent_apikey)
        self.__Agent_role = agent_role

    def LLM_model(self):

        print("fligt book details :")
        print(f"Agent Rool : {self.__Agent_role}")
        self.Agent_info()

class Hotel_booking(Agent):

    def __init__(self,agent_name,agent_apikey,agent_role):
        super().__init__(agent_name,agent_apikey)
        self.__agent_role =  agent_role
 
    def LLM_model(self):

        print(" Hotel_booking details:")
        print(f"Agent Role : {self.__agent_role}")
        self.Agent_info()
        
        

fly = flight_book("Opent AI",12345,"Flight booking")
hol = Hotel_booking("Gemini",2345,"Hotel booking")

fly.LLM_model()
print("#"*50)
hol.LLM_model()



    

```
