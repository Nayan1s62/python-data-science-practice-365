# encapsulation  is a fundamental principle in object oriented programming that focuses on building data and the methods that operate on the data into single unit called a class 

```python
# Access Modifier---->Encapsulation
# private  # 2 __ for it

class person:
    # constructor 
    def __init__(self,name,age):
        self.__name = name
        self.__age = age

    def display_info(self):
        print(f"the person name is {self.__name}  and age is {self.__age}")
        
```

```python
# private variable ko acces krna class ke andr hi thik hai baahar nai  
# private attribute use for hidden the attribute and keyword 
person = person("Nayan",22)
person.display_info()
```

```python
# Access Modifier---->Encapsulation
# protected - 1 _ under score for it 
# only used in derived class not in out of class 


class person:
    # constructor 
    def __init__(self,name,age):
        self._name = name
        self._age = age

    def display_info(self):
        print(f"the person name is {self._name}  and age is {self._age}")
        
```

```python
person = person("nayan",22)
```

```python
class Student(person):
    def __init__(self,name,age):
        super().__init__(name,age)
        
    def display_info(self):
        print(f"the person name is {self._name}  and age is {self._age}")
        
```
