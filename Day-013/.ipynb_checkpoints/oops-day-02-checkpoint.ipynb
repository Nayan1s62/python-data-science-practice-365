```python
import random

class BankAccount:
    """Blueprint for individual client accounts."""
    
    def __init__(self, account_holder, initial_deposit):
        # Attributes unique to each object
        self.account_holder = account_holder
        self.__balance = initial_deposit  # Private variable for data security
        self.account_number = random.randint(10000, 99999)
        
    def deposit(self, amount):
        """Adds money to the private balance framework."""
        if amount > 0:
            self.__balance += amount
            print(f"✓ ${amount} deposited successfully.")
        else:
            print("✗ Deposit amount must be positive.")
            
    def withdraw(self, amount):
        """Deducts money after validating funds."""
        if amount > self.__balance:
            print("✗ Transaction declined: Insufficient funds.")
        elif amount <= 0:
            print("✗ Withdrawal amount must be positive.")
        else:
            self.__balance -= amount
            print(f"✓ ${amount} withdrawn successfully.")
            
    def display_balance(self):
        """Safely exposes data via a getter function."""
        return self.__balance


class Bank:
    """Central manager to hold and operate all Account objects."""
    
    def __init__(self, bank_name):
        self.bank_name = bank_name
        self.accounts = {}  # Dictionary storing format: {account_number: BankAccount_Object}

    def open_account(self, name, deposit):
        """Creates a new BankAccount object and stores it."""
        if deposit < 0:
            print("✗ Cannot open an account with a negative balance.")
            return
        
        new_account = BankAccount(name, deposit)
        self.accounts[new_account.account_number] = new_account
        print(f"\n=== Account Created Successfully ===")
        print(f"Holder: {new_account.account_holder}")
        print(f"Account ID: {new_account.account_number}")
        print(f"Opening Balance: ${new_account.display_balance()}")

    def get_account(self, account_number):
        """Retrieves a specific object using its key identifier."""
        return self.accounts.get(account_number, None)


# --- Simulation / Main Driver Loop ---
if __name__ == "__main__":
    # Instantiating the primary Bank management object
    my_bank = Bank("Apex National Bank")
    
    # 1. Creating (instantiating) separate Account Objects
    my_bank.open_account("Alice Smith", 500.0)
    my_bank.open_account("Bob Jones", 1500.0)
    
    print("\n--- Processing Transactions ---")
    # 2. Accessing a specific account object dynamically
    target_id = list(my_bank.accounts.keys())[0]  # Grab Alice's generated ID
    alice_account = my_bank.get_account(target_id)
    
    if alice_account:
        # 3. Interacting with the object's methods
        alice_account.deposit(250)
        alice_account.withdraw(100)
        print(f"Final Balance for {alice_account.account_holder}: ${alice_account.display_balance()}")
```

```python
# example form anathoer 
# Employee Management

class Employee:

    # Constructor
    def __init__(self, emp_id, salary, pf_money):
        self.emp_id = emp_id
        self.__salary = salary          # Private Attribute
        self.__pf_money = pf_money      # Private Attribute

    # Getter Method
    def get_salary(self):
        return self.__salary

    def get_pf_money(self):
        return self.__pf_money

    # Setter Method for Salary
    def set_salary(self, salary):

        if salary > 10000:
            self.__salary = salary
            print("Salary Updated Successfully!")
        else:
            print("Salary must be greater than 10000.")

    # Setter Method for PF Money
    def set_pf_money(self, pf_money):

        if pf_money > 1000:
            self.__pf_money = pf_money
            print("PF Money Updated Successfully!")
        else:
            print("PF Money must be greater than 1000.")

    # Display Method
    def display_details(self):

        print("Employee ID :", self.emp_id)
        print("Employee Salary :", self.__salary)
        print("Employee PF Money :", self.__pf_money)


# Create Object
emp = Employee(101, 25000, 2000)

# Display Details
emp.display_details()

# Getter Method
print("\nSalary using Getter :", emp.get_salary())
print("PF Money using Getter :", emp.get_pf_money())

# Setter Method
emp.set_salary(30000)
emp.set_pf_money(2500)

print("\nAfter Updating")
emp.display_details()
```
