```python
import seaborn as sns
import matplotlib.pyplot as plt 
import pandas as pd 
import numpy as np
```

```python
#dataset store in this  
tips = sns.load_dataset("tips")
# irirs  Dataset 
iris = sns.load_dataset("iris")

titanic = sns.load_dataset("titanic")
# penguins 
penguin =  sns.load_dataset("penguin")

# flights 
flights = sns.load_dataset("flights")
```

```python
# Make all plots
# Create 2 x 4 grid
fig, axes = plt.subplots(2, 4, figsize=(20, 10))

# Title of figure
fig.suptitle(
    "Categorical Plots - All Types",
    fontsize=18,
    fontweight="bold"
)

# Convert 2D axes array into 1D array
axes = axes.flatten()

# 1. Strip Plot
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    
    palette="dark",
    ax=axes[0]
)

axes[0].set_title("Strip Plot")

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Theme
sns.set_theme(style="whitegrid")

# Create 2 x 4 grid
fig, axes = plt.subplots(2, 4, figsize=(20, 10))

# Main title
fig.suptitle(
    "Categorical Plots - All Types",
    fontsize=18,
    fontweight="bold"
)

# Convert 2D array into 1D
axes = axes.flatten()

# 1. Strip Plot
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    palette="dark",
    ax=axes[0]
)

# 2 sns.barplot()sns.countplot()sns.boxplot()sns.violinplot()sns.stripplot()sns.swarmplot()

# 2 barplot 
sns.barplot(data = tips,x = "day",y ="total_bill",hue="sex",palette ="dark",ax=axes[1])
# 3 violinplot
sns.violinplot(data = tips,x = "day",y ="total_bill",hue="sex",palette ="dark",ax=axes[2])
#4.swarmplot()
sns.swarmplot(data = tips,x = "day",y ="total_bill",hue="sex",palette ="dark",ax=axes[3])
# 5 sns.boxplot()
sns.boxplot(data = tips,x = "day",y ="total_bill",hue="sex",palette ="dark",ax=axes[4])
# .6 Count Plot
sns.countplot(
    data=tips,
    x="day",
    hue="sex",
    palette="dark",
    ax=axes[5]
)
# pointplot()
sns.pointplot(data = tips ,x ="day",y = "total_bill",hue ="sex",palette ="dark",ax=axes[6])

axes[0].set_title("Strip Plot")
axes[1].set_title("bar plot ")
axes[2].set_title("violin_plot")
axes[3].set_title("swarm_plot")
axes[4].set_title("box_plot")
axes[5].set_title("count_plot")
axes[6].set_title("point_plot")



plt.tight_layout()
plt.show()
```
