# Clean the Price Column Because of commas, Pandas may treat it as an object.

```python
df["price"] = df["price"].astype(str)
df["price"]
```

```python
df["price"] = df["price"].str.replace(",","")
df["price"].head(10)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Clean price column
df["price"] = (
    df["price"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
)

# Convert to numeric
df["price"] = pd.to_numeric(
    df["price"],
    errors="coerce"
)

# Create histogram
plt.hist(
    df["price"].dropna(),
    bins=20
)

plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")

plt.show()
```
