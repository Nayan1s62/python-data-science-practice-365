```python
import pandas as pd
import matplotlib.pyplot as plt

# Clean price column
df["price"] = (
    df["price"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
)

# Convert to numeric
df["price"] = pd.to_numeric(
    df["price"],
    errors="coerce"
)

# Create histogram
plt.hist(
    df["price"].dropna(),
    bins=20
)

plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")

plt.show()
```

# Add a grid

```python
plt.hist(df["price"].dropna(), bins=20)

plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")

plt.grid()

plt.show()
```

```python
plt.boxplot(
    df["price"].dropna()
)

plt.title("Price Box Plot")
plt.ylabel("Price")

plt.show()
```

```python
# Point Plot
```

```python
df.columns
```

```python
sns.pointplot(
    data = df,
    x = "brand",
    y = "price")
plt.title("Average laptop price by brand")
plt.xticks(rotation=45)
plt.show()
```

```python
top_brands = df["brand"].value_counts().head(10).index

top_df = df[df["brand"].isin(top_brands)]

plt.figure(figsize=(12, 6))

sns.pointplot(
    data=top_df,
    x="brand",
    y="price"
)

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```
