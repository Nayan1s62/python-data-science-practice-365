```python
import pandas as pd 
df = pd.read_csv("C:\\Users\\NAYAN\\Downloads\\Task_dataset.csv")
df["AnnualIncome"] = df["AnnualIncome"].fillna(df["AnnualIncome"].median())
df["Balance"] =df["Balance"].fillna(df["Balance"].mean())
df["Age"] = df["Age"].fillna(df["Age"].median())
print(df.describe())
```

```python
# challenge 8 – ML Dataset Preparation
```

```python
import pandas as pd

df = pd.read_csv("C:\\Users\\NAYAN\\Downloads\\Task_dataset.csv")

df = df.drop_duplicates()

df = df.drop(columns=[" CustomerID ","Segment"])

df["Age"] = df["Age"].fillna(df["Age"].median())
df["AnnualIncome"] = df["AnnualIncome"].fillna(df["AnnualIncome"].median())
df["Balance"] = df["Balance"].fillna(df["Balance"].mean())

df = df.rename(columns={"Target":"Output"})

print(df.head())

df.info()

print(df.describe())

print(df["Output"].value_counts())
```
