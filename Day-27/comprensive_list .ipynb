```python
text = input()

frequency = {}

for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

for char, count in frequency.items():
    print(char + ":" + str(count), end=" ")
```

```python
text = input()

count = 0

for char in text:
    if char.isdigit():
        count += 1

print(count)
```

```python
sentence = input()

words = sentence.split()

longest = ""

for word in words:
    if len(word) > len(longest):
        longest = word

print(longest)
```
