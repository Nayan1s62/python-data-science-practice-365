OOP Problems

Create a basic class

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def show(self):
        print(f"Name: {self.name}, Marks: {self.marks}")

s1 = Student("Nayan", 88)
s1.show()
```

2. Constructor with default values

```python
class Car:
    def __init__(self, brand, color="White"):
        self.brand = brand
        self.color = color

    def display(self):
        print(f"{self.color} {self.brand}")

car1 = Car("Toyota")
car2 = Car("BMW", "Black")

car1.display()
car2.display()
```

3. Instance vs class variables

```python
class Employee:
    company = "TCS"          # class variable
    employee_count = 0

    def __init__(self, name):
        self.name = name      # instance variable
        Employee.employee_count += 1

e1 = Employee("Amit")
e2 = Employee("Priya")

print(Employee.company)
print(Employee.employee_count)
```

4. Encapsulation with private variable

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance   # private variable

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Invalid withdrawal")

    def get_balance(self):
        return self.__balance

account = BankAccount("Nayan", 1000)
account.deposit(500)
account.withdraw(300)

print(account.get_balance())
```

5. Inheritance

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating")

class Dog(Animal):
    def bark(self):
        print(f"{self.name} says Woof!")

dog = Dog("Bruno")
dog.eat()
dog.bark()
```

6. Method overriding

```python
class Shape:
    def area(self):
        print("Area not defined")

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        print(f"Circle area: {3.14 * self.radius ** 2}")

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        print(f"Square area: {self.side ** 2}")

shapes = [Circle(5), Square(4)]

for shape in shapes:
    shape.area()
```

Polymorphism

```python
class Dog:
    def sound(self):
        print("Dog barks")

class Cat:
    def sound(self):
        print("Cat meows")

class Cow:
    def sound(self):
        print("Cow moos")

animals = [Dog(), Cat(), Cow()]

for animal in animals:
    animal.sound()
```
