```python
import matplotlib.pyplot as plt
import numpy as  np
```

```python
X = [1,2,3,4,5]
Y = [10,11,12,13,14]
plt.plot(X,Y)
plt
```

```python
# i want to some labeling and design part 
plt.plot(X,Y,color ="red",linewidth=2,marker = "*",markersize = 10,markeredgecolor="blue",markerfacecolor = "green")
plt.title("first plot of nayan")
plt.xlabel("X Axis ")
plt.ylabel("Y Axis")
plt.show()
```

```python
# pyplot common commands 
X = np.linspace(0,10,100)

Y = np.sin(X)

X
plt.plot(X,Y,color ="red",linewidth=2,marker = "*",markersize = 10,markeredgecolor="purple",markerfacecolor = "black")
plt.title("first plot of nayan")
plt.xlabel("X Axis ")
plt.ylabel("Y Axis")
plt.show()
# commands
plt.plot(X,Y,color ="red",linewidth=2,marker = "*",markersize = 10,markeredgecolor="purple",markerfacecolor = "black")
plt.title("sine Waves")
plt.xlabel("X Axis data")
plt.ylabel("Y axis daata")
plt.xlim(1,10)
plt.ylim(-1.5,1.5)
plt.xticks([0,2,4,6,8,10])
plt.yticks([-1,0,1])
```

```python
plt.plot(X,Y,color ="red",linewidth=2,marker = "*",markersize = 10,markeredgecolor="purple",markerfacecolor = "black")
plt.title("sine Waves")
plt.xlabel("X Axis data")
plt.ylabel("Y axis daata")
plt.xlim(1,10)   # x range 
plt.ylim(-1.5,1.5) # y range 
plt.xticks([0,2,4,6,8,10]) # set  x  tick marks  
plt.yticks([-1,0,1])      # set y tick marks 
plt.grid(True, alpha=0.9)  # only 0-1 # show grid 
plt.legend(["sin X waves "]) # add the legend showing the information on the graph 
plt.tight_layout() # autofix spaceing
plt.show()

```
