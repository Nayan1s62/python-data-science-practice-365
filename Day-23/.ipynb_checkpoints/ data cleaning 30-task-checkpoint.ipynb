```python
#2. Display the number of rows and columns in the dataset.
df.shape
```

```python
#3. Display all column names and identify which column names have
#leading/trailing spaces.
df.columns.tolist()
for i in df.columns :
    if i != i.strip():
        print(f"space found ",repr(i)) #  repr ka use kr ke spaces ko clear ly visible dekh skte hai hum 
    
```

```python
#4. Clean all column names by removing extra spaces and converting them into a
#consistent format such as snake_case.
df.columns.tolist()
for i in df.columns :
    if (i != i.strip()) & (i!= i.replace("","_")):
        print(f"Space found and removed",repr(i))

            
        

    
```
