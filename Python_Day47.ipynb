Write a Python program to find the roots of a quadratic equation of the form:ax2+bx+c=0
The program should:
Accept values of a, b, and c as input from the user.
Calculate the discriminant using the formula:D = b² − 4ac
Use the quadratic formula to compute the roots.
Display both roots of the equation.

```python
import math

a = float(input())
b = float(input())
c = float(input())

D = b**2 - 4*a*c

root1 = (-b + math.sqrt(D)) / (2*a)
root2 = (-b - math.sqrt(D)) / (2*a)

print("Roots are:", max(root1, root2), "and", min(root1, root2))
```

```python
Write a Python program to accept a string from the user.
Count how many times each character appears in the string.
Store the character counts in a dictionary.
Print the dictionary showing each character and its frequency.
```

```python
# Accept a string from the user
text = input()

# Create an empty dictionary
frequency = {}

# Count each character
for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

# Print the character frequencies
print(frequency)
```

```python
Write a Python program to check the strength of a password.
A password is strong if it has at least 8 characters, one uppercase letter, one digit, and one special character.
Use conditions to verify all these requirements.
Print "Strong Password" if all conditions are satisfied, otherwise print "Weak Password".
```

```python
password = input()

has_uppercase = any(char.isupper() for char in password)
has_digit = any(char.isdigit() for char in password)
has_special = any(not char.isalnum() for char in password)

if len(password) >= 8 and has_uppercase and has_digit and has_special:
    print("Strong Password")
else:
    print("Weak Password")
```
