# Write a Python program to read a number.
 Check if the number lies between 0 and 9.
 Print Yes if it is a single digit.
 Otherwise, print No.

```python
 
num = int(input())

if 0 <= num <= 9:
    print("Yes")
else:
    print("No")
```
