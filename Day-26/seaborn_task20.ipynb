```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# importing the data set in to tables 
titanic = sns.load_dataset("titanic")
titanic
```

# 1. Survival by Gender
# Use catplot() to show the number of passengers who survived and did not survive for each
# gender.
# • X-axis → sex
# • Hue → survived
# • Chart → count

```python
sns.catplot(
    data=titanic,
    x="survived",
    hue = "sex",
    kind= "count")
plt.title("Survival by Gender")
plt.xlabel("genders ")
plt.ylabel("survival count")

plt.show()
```

 # 2. Survival by Passenger Class
# Create a categorical plot showing the number of passengers in each class.
#Use:
#• x="class"
#• kind="count"

```python
sns.catplot(
    data=titanic,
    x="class",
    kind="count")
plt.title("Survival by Passenger Class")
plt.xlabel("class")


plt.show()
```

```python
#4 Create a categorical visualization comparing survival across sex.
#Use catplot() and choose an appropriate categorical plot.

# Then identify which gender has a higher survival rate.

```

```python
sns.catplot(
    data=titanic,
    x="sex",
    y="survived",
    kind="bar"
)

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")

plt.show()
```
