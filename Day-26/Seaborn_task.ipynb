```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create figure
plt.figure(figsize=(10, 8))

# Box plot
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="smoker",
    palette=["lightblue", "lightgreen"],
    width=0.7,
    linewidth=1.5,
    flierprops={
        "marker": "o",
        "markerfacecolor": "red",
        "markeredgecolor": "black",
        "markersize": 6,
        "alpha": 0.7
    }
)

# Days
days = ["Thur", "Fri", "Sat", "Sun"]

# Calculate median for each day
median = [
    tips[tips["day"] == d]["total_bill"].median()
    for d in days
]

# Add annotation
for i, value in enumerate(median):

    plt.annotate(
        f"Median: {value:.2f}",
        xy=(i, value),
        xytext=(i, value + 2),
        ha="center",
        fontsize=10,
        fontweight="bold",
        arrowprops={
            "arrowstyle": "->"
        }
    )

plt.title("Total Bill by Day and Smoker")
plt.xlabel("Day")
plt.ylabel("Total Bill")

plt.show()
```
