# line plot with the relplot 

```python
flights = sns.load_dataset("flights")
# make the plot 
sns.relplot(data = flights,
           x = 'year',
           y= 'passengers',
           hue = 'month', # divide the dataset 
           kind = 'line',
           col='month')    # single colum in diffrent graph 
```

```python
flights = sns.load_dataset("flights")
# make the plot 
sns.relplot(data = flights,
           x = 'year',
           y= 'passengers',
           hue = 'month', # divide the dataset 
           kind = 'line',
           col='month',
           col_wrap= 2, # 4 column per row 
           palette ='tab20',
            height= 3,linewidth = 2,
           aspect =1.2)    # col_wrap is used in colum data graph 
# single colum in diffrent graph 
plt.title("line plot month wise ")
plt.show()
```

# regression plot # regplot() function it draws straight line between the scatter plot it called as a regression line 


```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    color="pink",
    scatter_kws={"alpha": 0.9, "color": "blue"},
    line_kws={"color": "red", "linewidth": 5, "alpha": 0.9}
)

plt.title("Regression Plot: Total Bill vs Tip")
plt.show()
```

```python
# ploynomial regression (order = 2 )
import seaborn as sns
import matplotlib.pyplot as plt

sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    order = 1,
    color="red",
    scatter_kws={"alpha": 0.9, "color": "blue"},
)

plt.title("Regression Plot: Total Bill vs Tip")
plt.show()
```
