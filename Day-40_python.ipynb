#  Write a Python program to read three sides of a triangle.
Check if all sides are equal, any two are equal, or all are different.
Print Equilateral, Isosceles, or Scalene.
Print only one triangle type.


```python
# Write your code here
a = int(input())
b = int(input())
c = int(input())

if a == b == c:
    print("Equilateral")
elif a == b or b == c or a == c:
    print("Isosceles")
else:
    print("Scalene")
```

Write a Python program to read two strings.
Compare their lengths using len() function.
Print which string is longer.
Print Same length if both are equal.

```python
str1 = input()
str2 = input()

if len(str1) > len(str2):
    print("First string is longer")
elif len(str2) > len(str1):
    print("Second string is longer")
else:
    print("Same length")
```

Write a Python program to read two decimal numbers.
Round both numbers to 2 decimal places.
Compare the rounded values.
Print Equal or Not Equal.

```python
a = float(input())
b = float(input())

a = round(a, 2)
b = round(b, 2)

if a == b:
    print("Equal")
else:
    print("Not Equal")
```
