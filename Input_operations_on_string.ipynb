Write a Python program to take a year as input.
Check whether the year satisfies leap year conditions.
A leap year occurs based on divisibility rules.
If it is a leap year, print Leap Year, otherwise print Not a Leap Year.

```python
year = int(input())

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```

Write a Python program to take three numbers as input.
Identify the largest number first.
Ignore the highest value and compare the remaining two.
Print the second largest number.


```python
a = int(input())
b = int(input())
c = int(input())

if a >= b and a >= c:
    largest = a
elif b >= a and b >= c:
    largest = b
else:
    largest = c

if a == largest:
    a = float('-inf')
if b == largest:
    b = float('-inf')
if c == largest:
    c = float('-inf')

if a >= b and a >= c:
    print(a)
elif b >= a and b >= c:
    print(b)
else:
    print(c)
```

Write a Python program to take marks of 5 subjects separately.
Calculate the total marks obtained.
Assign grade A if total is greater than 90, B if greater than 80.
Assign C if greater than 70, D if greater than 60, else F.


```python
m1 = int(input())
m2 = int(input())
m3 = int(input())
m4 = int(input())
m5 = int(input())

total = m1 + m2 + m3 + m4 + m5

if total > 90:
    print("A")
elif total > 80:
    print("B")
elif total > 70:
    print("C")
elif total > 60:
    print("D")
else:
    print("F")
```
