```python
class Numbers:

    def __iter__(self):
        self.num = 1
        return self

    def __next__(self):
        if self.num <= 10:
            value = self.num
            self.num += 1
            return value
        else:
            raise StopIteration

obj = Numbers()

for i in obj:
    print(i)
```

```python
class Reverse:

    def __iter__(self):
        self.num = 10
        return self

    def __next__(self):
        if self.num >= 1:
            value = self.num
            self.num -= 1
            return value
        else:
            raise StopIteration

obj = Reverse()

for i in obj:
    print(i)
```

```python
class Word:

    def __iter__(self):
        self.word = "GENERATOR"
        self.index = 0
        return self

    def __next__(self):
        if self.index < len(self.word):
            ch = self.word[self.index]
            self.index += 1
            return ch
        else:
            raise StopIteration

obj = Word()

for i in obj:
    print(i)
```

```python
class Table:

    def __iter__(self):
        self.num = 1
        return self

    def __next__(self):
        if self.num <= 10:
            value = 5 * self.num
            self.num += 1
            return value
        else:
            raise StopIteration

obj = Table()

for i in obj:
    print(i)
```

```python
def numbers():
    for i in range(1, 11):
        yield i

for i in numbers():
    print(i)
```

```python
numbers = [10, 15, 20, 25, 30, 35]

def even():
    for i in numbers:
        if i % 2 == 0:
            yield i

for i in even():
    print(i)
```

```python
numbers = [1, 4, 7, 10, 13, 16, 19]

def odd():
    for i in numbers:
        if i % 2 != 0:
            yield i

for i in odd():
    print(i)
```

```python
names = ["Alice", "Bob", "Charlie", "David", "Emma"]

def name_generator():
    for name in names:
        yield name

for i in name_generator():
    print(i)
```

```python
numbers = [25, 60, 45, 80, 55, 30, 90]

def greater():
    for i in numbers:
        if i > 50:
            yield i

for i in greater():
    print(i)
```

```python
class Fibonacci:

    def __iter__(self):
        self.a = 0
        self.b = 1
        self.count = 0
        return self

    def __next__(self):
        if self.count < 10:
            value = self.a
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return value
        else:
            raise StopIteration

obj = Fibonacci()

for i in obj:
    print(i)
```

```python
def fibonacci():
    a = 0
    b = 1

    for i in range(15):
        yield a
        a, b = b, a + b

for i in fibonacci():
    print(i)
```

```python
def prime():

    for num in range(2, 101):

        is_prime = True

        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            yield num

for i in prime():
    print(i)
```

```python
num = 24

def factors():

    for i in range(1, num + 1):
        if num % i == 0:
            yield i

for i in factors():
    print(i)
```

```python
def countdown():

    for i in range(20, 0, -1):
        yield i

for i in countdown():
    print(i)
```

```python
def multiples():

    num = 5

    while True:
        yield num
        num += 5

gen = multiples()

for i in range(15):
    print(next(gen))
```
