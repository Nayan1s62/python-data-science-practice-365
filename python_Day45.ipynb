Write a Python program to read an integer.
Add 1 to the number without using the + operator.
Use only bitwise operations to perform addition.
Ensure the value increases correctly by one.
Print the result.

```python
# Read a non-negative integer from the user
n = int(input(" "))

original_n = n
mask = 1

# Bitwise addition of 1 (simulating carry propagation)
while (n & mask) != 0:
    n = n ^ mask
    mask = mask << 1

n = n ^ mask

print(n)
```

Write a Python program to read an integer.
Convert the given number into its binary form manually.
Do not use Python’s built-in bin() function.
Use bitwise right shift operations to extract each bit.
Print the binary representation of the number.

```python
# Read a non-negative integer from the user
n = int(input("Enter a non-negative integer: "))

if n == 0:
    binary_representation = "0"
else:
    binary_representation = ""
    temp = n
    
    # Loop until all bits have been shifted out
    while temp > 0:
        # Extract the least significant bit using bitwise AND
        bit = temp & 1
        
        # Prepend the bit to the binary string
        binary_representation = str(bit) + binary_representation
        
        # Right shift by 1 to process the next bit
        temp = temp >> 1

print(f"The binary representation of {n} is: {binary_representation}")
```

Write a program that takes a number and a position value i from the user
Clear all bits from position 0 (LSB) up to position i in that number
All bits beyond position i should remain the same
Display the final result after clearing the bits
This uses bitwise operations to modify specific bits in a number


```python
# Read the number and position from the user
n = int(input(" "))
i = int(input(" "))

# Create a bitwise mask:
# -1 has all bits set to 1 (...111111).
# Shifting it left by (i + 1) places 0s from position 0 to i, leaving 1s above.
mask = -1 << (i + 1)

# Apply the mask using the bitwise AND operator
result = n & mask


print(result)
```

Write a Python program to read an 8-bit number (0–255).
Flip every bit of the number using bitwise operations.
Change all 0 bits to 1 and all 1 bits to 0.
Ensure the result remains within 8-bit range.
Print the final flipped value.


```python
# Read an 8-bit integer from the user
n = int(input(" "))

if 0 <= n <= 255:
    # 1. Use bitwise NOT (~) to flip the bits
    # 2. Use '& 0xFF' (an 8-bit mask of all 1s) to restrict the result to the 8-bit range (0-255)
    flipped_value = ~n & 0xFF
    
   
    print(flipped_value )
else:
    print("Error: Please enter a number strictly between 0 and 255.")
```
