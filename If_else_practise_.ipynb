**Write a Python program to check whether a number is prime. Accept an integer from the user. Use a for loop to check divisibility of the number. Determine whether the number has factors other than 1 and itself. Print whether the number is prime or not.**

####

```python
num = int(input("Enter a number: "))

if num <= 1:
    print("Not a Prime Number")
else:
    for i in range(2, num):
        if num % i == 0:
            print("Not a Prime Number")
            break
    else:
        print("Prime Number")
```

```python
Write a Python program to calculate the sum of digits of a number.
Take an integer input from the user.
Use a for loop to process each digit.
Add each digit to a running total.
Print the final sum of digits.
```

```python
num = input("Enter a number: ")

total = 0

for digit in num:
    total = total + int(digit)

print("Sum of digits:", total)
```

```python
**Write a Python program to print the Fibonacci series up to n terms. Accept the number of terms as input from the user. Initialize the first two Fibonacci numbers. Use a for loop to generate the remaining terms. Print the Fibonacci sequence in order.**
```

```python
n = int(input("Enter the number of terms: "))

a = 0
b = 1

for i in range(n):
    print(a, end=" ")
    c = a + b
    a = b
    b = c
```
